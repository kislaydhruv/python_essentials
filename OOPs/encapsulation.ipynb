```python
class BankAccount:
    def __init__ (self,owner,balance):
        self.owner = owner 
        self.__balance = balance 

    def deposit(self , amount):
        if amount > 0 :
            self.__balance += amount

    def get_balance(self):
        return self.__balance
```

```python
 acc = BankAccount('Alice',5000)
 acc.deposit(1000)
```

```python
print(acc.get_balance())
```

### getters and setters


```python
class Temperature :
    def __init__ (self,celcius):
        self.__c = celcius
    @property
    def celcius(self):
        return self.__c

    @celcius.setter
    def celcius(self, v):
        if v < -273.15 : raise ValueError ('Below absolute zero! ')
        self.__c = v

    
    @property
    def fahrenheit(self):
        return self.__c * 9/5 + 32
```

```python
t = Temperature(25)
print(t.fahrenheit)
t.celcius = 100
print(t.fahrenheit)
```
