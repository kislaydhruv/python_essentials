```python
class Animal:
    def __init__ (self,name): # describing a constructor 
        self.name = name 
    def speak(self): # describing a method 
        print(f'{self.name} makes a sound. ')
```

```python
class Dog(Animal):
    def __init__(self,name): # describing a method
        super().__init__(name) # call parent constructor 
    def speak(self):
        print(f'{self.name} says woof! ') # override the parent method 
    def fetch(self):
        print(f'{self.name} fetches!')
```

```python
d = Dog('Buddy')
d.speak()
d.fetch()
```
