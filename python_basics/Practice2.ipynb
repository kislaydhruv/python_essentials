## Functions


### Task 1

```python
def apply_discount(price , discount_percent=5):
    return price - (price*discount_percent)/100
```

```python
print(apply_discount(1000,10))
print(apply_discount(500))
```

### TASK 2

```python
def factorial(n):
    if(n<0):
        print("It's an error")
        return 
    elif(n==0 or n==1):
        return 1 
    
    return n*factorial(n-1)
```

```python
print(factorial(5))
```

```python
print(factorial(0))
```

```python
print(factorial(-3))
```
