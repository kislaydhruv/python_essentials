## for Loop


```python
for fruit in ['apple','banana','cherry']:
    print(fruit)
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(2,10,2):
    print(i)
```

```python
for i,val in enumerate (['a','b','c']):
    print(i,val)
```

```python
names = ['Alice', 'Bob']
scores = [90, 85]
for name, score in zip(names, scores):
 print(f'{name}: {score}')
```

## While

```python
count = 0
while count < 5:
  print(count)
  count += 1  
```
